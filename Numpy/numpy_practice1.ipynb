```python
# 6.) Write a program to create the following 1D array:
#  [1, 2, 3, 4, 5]
# Then print:
# - Array
# - Shape
# - Number of dimensions

import numpy as np

arr = np.array([1,2,3,4,5])

print(arr)

print(f"Shape of the Array : {arr.shape}")

print(f"Number of Dimensions : {arr.ndim}")
```
