# 1.) What is Vectorization?

Imagine you are a teacher.

You have marks of 10,000 students.

You want to add 5 bonus marks to every student.

There are two ways:

Method 1 (Python List)

Go student by student.

Student 1 → +5
Student 2 → +5
Student 3 → +5
...
Student 10000 → +5

This means Python has to repeat the same work again and again.

----------------------------

Method 2 (NumPy)

Tell NumPy only once:

"Add 5 to everyone."

NumPy performs the operation on the entire array at once.

This idea is called Vectorization.

-------------------

Definition:
Vectorization means performing an operation on an entire NumPy array at once without writing explicit Python loops.

---------------------

# Why does Vectorization exist?

Suppose we have

 - 1 million numbers

If Python processes them one by one, it becomes slower.

- NumPy was created for scientific computing.

- Instead of asking Python to repeat millions of operations,

- NumPy lets optimized C code do the work internally. This is one reason NumPy is much faster for numerical computations.

# Who predefined Vectorization?

You never write a function called vectorization().

 - It is not a function.

 - It is a feature of the NumPy library.

The developers of NumPy designed its array operations so that expressions like:

 - array + 5

or

 - array * 2

automatically work on every element.
